```python
from dbiolab.mammalian_cellculture_simulation import (
    CellType,
    FeedConcentrations,
    FeedingEvent,
    MammalianCellcultureReactor,
    MammalianRecipe,
    NoiseParameters,
    ReactorType,
    plot_reaction_evolution_timeseries,
)
```

```python
noise_parameters = NoiseParameters(
    process_noise_c_product=0.02,
    process_noise_x_total=0.02,
    process_noise_x_viable=0.02,
    process_noise_x_dead=0.02,
    process_noise_c_glc=0.02,
    process_noise_c_gln=0.02,
    process_noise_c_lac=0.02,
    process_noise_c_ammonia=0.02,
    process_noise_c_aggregates=0.02,
    measurement_noise_c_product=0.01,
    measurement_noise_x_total=0.01,
    measurement_noise_x_viable=0.01,
    measurement_noise_x_dead=0.01,
    measurement_noise_c_glc=0.01,
    measurement_noise_c_gln=0.01,
    measurement_noise_c_lac=0.01,
    measurement_noise_c_ammonia=0.01,
    measurement_noise_c_aggregates=0.01,
)
```

```python
no_noise_parameters = NoiseParameters()
```

```python
reactor = ReactorType.from_json("../.local/MTP_1.json")
simulator = MammalianCellcultureReactor(reactor_type=reactor)
celltype = CellType.from_json("../.local/celltype_1_alpha.json")
initial_conditions = FeedConcentrations.from_json("../.local/initial_conditions.json")
feed_event = FeedingEvent.from_json("../.local/feed_event.json")
feeding_strategy = [feed_event]
recipe = MammalianRecipe(
    name="Test Recipe",
    celltype=celltype,
    temperature=37.4,  # range 30-40
    ph_value=6.6,  # range 6-8
    initial_conditions=initial_conditions,
    feeding_strategy=feeding_strategy,
    runtime=1336.0,
    measurement_intervals=12.0,
    noise_parameters=no_noise_parameters,
)
n_samples = 1
cells = []
metabolites = []
products = []
purity = []
for i in range(n_samples):
    solution = simulator.run_experiment(recipe=recipe, seed=i)
    figures = plot_reaction_evolution_timeseries(solution)
    cells.append(figures[0])
    metabolites.append(figures[1])
    products.append(figures[2])
    purity.append(figures[3])
```

```python
for i in range(n_samples):
    cells[i].show()
```
