```python
import json
from pathlib import Path

from dbiolab.mammalian_cellculture_simulation import (
    ReactorType,
    CellType,
    MammalianCellcultureReactor,
    MammalianRecipe,
    FeedConcentrations,
    FeedingEvent,
)
from dbiolab.visualisation.plot_reaction_timeseries import (
    plot_reaction_evolution_timeseries,
)
```

```python
reactor = ReactorType.from_json(
    "../src/dbiolab/mammalian_cellculture_simulation/reactor_types/MTP_1.json"
)
simulator = MammalianCellcultureReactor(reactor_type=reactor)

celltype = CellType.from_json(
    "../src/dbiolab/mammalian_cellculture_simulation/celltypes/craven_3L_batch.json"
)
initial_conditions = FeedConcentrations(
    c_product=0.0,
    x_total=0.4 * 1e6,
    x_viable=0.4 * 1e6,
    x_dead=0.0,
    c_glc=2.0,
    c_gln=4.0,
    c_lac=0.0,
    c_ammonia=1.8,
)
feed_event = FeedingEvent(
    time_point=200.0,
    feed_concentrations=FeedConcentrations(
        c_product=0.0,
        x_total=0.0,
        x_viable=0.0,
        x_dead=0.0,
        c_glc=2.0,
        c_gln=2.0,
        c_lac=0.0,
        c_ammonia=0.0,
    ),
)
feeding_strategy = []
recipe = MammalianRecipe(
    name="Test Recipe",
    celltype=celltype,
    temperature=38.5,  # range 30-40
    ph_value=7.5,  # range 6-8
    initial_conditions=initial_conditions,
    feeding_strategy=feeding_strategy,
    runtime=1336.0,
    measurement_intervals=12.0,
)

solution = simulator.run_experiment(recipe=recipe)
solution_plots = plot_reaction_evolution_timeseries(solution)
for fig in solution_plots:
    fig.show()
```
